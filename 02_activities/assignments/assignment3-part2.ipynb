 > What software did you use to create your data visualization?
 Rstudio 

> Who is your intended audience? 
Residents and policy makers interested in where Cafe's are distributed across Toronto
    
> What information or message are you trying to convey with your visualization? 
The distibution of cafes across the city
    
> What aspects of design did you consider when making your visualization? How did you apply them? With what elements of your plots? 
Whether the plot can be interpreted, I overlayed the heat map on an actual map of Toronto so viewers can discern where the cafes actually are in the city.
    
> How did you ensure that your data visualizations are reproducible? If the tool you used to make your data visualization is not reproducible, how will this impact your data visualization? 
The data is publically accessible and the information used in the graph is all available in the csv. The hardest part to reproduce is the map as that was added to the data but the coordinates are universal so a the data could be overlayed on any map and look comparable.
    
> How did you ensure that your data visualization is accessible?  
I used virdis to ensure the colours are able to be seen for those with and without visual impairments
    
> Who are the individuals and communities who might be impacted by your visualization?  
Those in populated areas who have less cafe's - policy makers will be able to make this comparison and see where services are lacking.
    
> How did you choose which features of your chosen dataset to include or exclude from your visualization? 
I included only location information. I could have provided more detail on what type of establishment it was (sidewalk, patio, or curblane) but I decided that this would reduce the legibility of the visualization and the most pertinant information was location.
    
> What ‘underwater labour’ contributed to your final data visualization product?
Collecting and updating the data requires underwater labour from CafeTO, additionally those who created the R packages to make heatmaps and accessibile colour palletes put in underwater labour.
